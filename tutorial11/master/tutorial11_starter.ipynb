## 1. Python Modules

### What is a Module?

A **module** in Python is simply a `.py` file that contains Python code (functions, classes, variables). Modules let you organize your code into separate files and reuse it across different programs.

Python comes with many built-in modules (like `math`, `random`, `csv`), and you can also create your own.

### Importing Modules

There are several ways to import a module:

**1. `import module_name`** — imports the entire module. You access its contents with `module_name.function_name()`.

```python
import math
print(math.pi)        # 3.141592653589793
print(math.sqrt(16))  # 4.0
```

**2. `from module_name import something`** — imports specific items directly into your namespace.

```python
from math import pi, sqrt
print(pi)        # 3.141592653589793
print(sqrt(16))  # 4.0
```

**3. `import module_name as alias`** — imports the module with a shorter name.

```python
import math as m
print(m.pi)  # 3.141592653589793
```

### Creating Your Own Module

To create your own module, simply write your functions in a `.py` file and place it in the **same folder** as your notebook or script. Then you can import it just like any other module.

For example, if you have a file called `math_helpers.py` in the same folder:

```python
# math_helpers.py
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def average(numbers):
    return sum(numbers) / len(numbers)
```

You can then import and use it:

```python
import math_helpers
print(math_helpers.add(3, 5))  # 8
```

## 2. CSV Module in Python

#### What is a CSV file?
- CSV stands for "Comma-Separated Values".
- A CSV file is a plaintext format where each line typically represents a row, and each value within the row is separated by a delimiter (commonly a comma).
- Commonly used for data exchange, simple data storage, and as input-output for many software applications.

#### Why use the `csv` module?
- The `csv` module in Python provides a simple interface to read and write CSV files.
- It handles splitting lines on the appropriate delimiter, which helps avoid manual string parsing errors.

#### Reading a CSV file with `csv.reader`

The `csv.reader` object takes an open file and returns an **iterable** — you can loop through it with a `for` loop. Each row comes back as a **list of strings**.

**Important:** All values from a CSV are **strings**. If you need numbers, you must convert them with `int()` or `float()`.

Below is an example of using the `csv` module to read from a file.

````python
# EXAMPLE: Reading from a CSV using the csv module
import csv

file = open('data.csv', mode='r', newline='')
csv_reader = csv.reader(file)
for row in csv_reader:
    print(row)
file.close()
````

#### Skipping the header row with `next()`

Often the first row of a CSV is a **header** (column names). You can skip it using `next()`:

```python
file = open('data.csv', mode='r', newline='')
csv_reader = csv.reader(file)
header = next(csv_reader)  # reads and skips the first row
```

#### Writing to a CSV File with `csv.writer`

`csv.writer` lets you write rows to a CSV file. Each row is a list, and the writer handles inserting commas for you.

**Important:** Always include `newline=''` when opening a file for CSV writing to avoid extra blank lines.

Remember to always call `.close()` on your file when you are done with it.

Below is a minimal example:

````python
import csv

data_to_write = [
    ["student_id", "exam1", "exam2", "exam3"],
    ["S101", "85", "90", "88"],
    ["S102", "78", "92", "80"],
]

# EXAMPLE: Writing to a CSV using the csv module
file = open('new_student_grades.csv', mode='w', newline='')
csv_writer = csv.writer(file)
for row in data_to_write:
    csv_writer.writerow(row)
file.close()

print("Data has been written to 'new_student_grades.csv'.")
````

## 4. Object-Oriented Programming (OOP)

#### Printing Objects in Python
- By default, printing an object like `print(my_object)` will show something like `<__main__.MyClass object at 0x...>`.
- To customize this, define the `__str__(self)` method inside the class. Then, `print()` will display the string you return from `__str__`.

````python
# EXAMPLE: Basic class WITHOUT __str__
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

car = Vehicle("Toyota", "Corolla")
print(car)  # Will print the default (ugly) string representation
````

````python
# EXAMPLE: Basic class WITH __str__
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def __str__(self):
        return "Vehicle (" + self.brand + ", " + self.model + ")"

car = Vehicle("Toyota", "Corolla")
print(car)  # Will use the __str__ method
````
